```python
from pathlib import Path

import nbformat
from nbconvert import MarkdownExporter, PDFExporter
from traitlets.config import Config


def ipynb_to_pdf(
    input_file,
    output_file,
):

    exporter = PDFExporter()
    nb = nbformat.read(input_file, as_version=4)
    pdf_bytes, _ = exporter.from_notebook_node(nb)
    with open(output_file, "wb") as f:
        f.write(pdf_bytes)


def ipynb_to_md(
    input_file,
    output_file,
):
    exporter = MarkdownExporter()
    nb = nbformat.read(input_file, as_version=4)
    md_text, _ = exporter.from_notebook_node(nb)
    with open(output_file, "w", encoding="utf-8") as f:
        f.write(md_text)
```

```python
for i in [1, 2, 3, 4, 5]:
    ipynb_to_pdf(
        f"/home/jovyan/mri/solutions/exercise-{i}-solution/exercise-{i}-solution.ipynb",
        f"/home/jovyan/mri/solutions/pdf/exercise-{i}-solution.pdf",
    )
```

```python
for i in [1, 2, 3, 4, 5]:
    ipynb_to_md(
        f"/home/jovyan/mri/solutions/exercise-{i}-solution/exercise-{i}-solution.ipynb",
        f"/home/jovyan/mri/solutions/md/exercise-{i}-solution.md",
    )
```
