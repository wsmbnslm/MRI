```python
import subprocess
```

```python
def md_to_pdf(input_path: str, output_path: str) -> None:
    """Konvertiert eine Markdown-Datei per pandoc + xelatex in ein PDF."""
    cmd = [
        "pandoc",
        input_path,
        "-o",
        output_path,
        "--pdf-engine=xelatex",
        "-V",
        "geometry:margin=2cm",
        "-V",
        "mainfont=DejaVu Serif",
        "-V",
        "monofont=DejaVu Sans Mono",
        "-V",
        "fontsize=10pt",
        "--syntax-highlighting=tango",
    ]
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode != 0:
        raise RuntimeError(f"pandoc-Fehler:\n{result.stderr}")
    print(f"PDF erstellt: {output_path}")
```
