```python
import os
import re

import nbformat
```

```python
def clear_solutions(input_nb, output_nb):

    nb = nbformat.read(input_nb, as_version=4)

    for cell in nb.cells:

        if cell.cell_type in ("code", "raw") and "# Your solution here" in cell.source:

            # Behalte NUR den exakten Kommentar und füge 3 leere Zeilen hinzu
            cell.source = "# Your solution here\n\n\n"

    os.makedirs(os.path.dirname(output_nb), exist_ok=True)

    nbformat.write(nb, output_nb)
```

```python
for i in range(3):
    clear_solutions(
        f"/home/jovyan/mri/solutions/exercise-{i+1}-solution/exercise-{i+1}-solution.ipynb",
        f"/home/jovyan/mri/exercises/exercise-{i+1}/exercise-{i+1}.ipynb",
    )
```
