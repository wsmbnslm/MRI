```python
import jupytext
import nbformat

def ipynb_to_md(
    input_file,
    output_file,
):
    nb = nbformat.read(input_file, as_version=4)
    jupytext.write(nb, output_file, fmt="md")
```

```python
for i in [1, 2]:
    ipynb_to_md(
        f"/home/jovyan/fti/solutions/exams-solution/mock-exam-{i}-solution.ipynb",
        f"/home/jovyan/fti/solutions/md/mock-exam-{i}-solution.md",
    )
```
