```python
import jupytext
import nbformat


def md_to_ipynb(
    input_file,
    output_file,
):
    nb = jupytext.read(input_file, fmt="md")
    nbformat.write(nb, output_file)
```

```python
for i in [1, 2]:
    md_to_ipynb(
        f"/home/jovyan/fti/solutions/md/mock-exam-{i}-solution.md",
        f"/home/jovyan/fti/solutions/exams-solution/mock-exam-{i}-solution.ipynb",
    )
```
