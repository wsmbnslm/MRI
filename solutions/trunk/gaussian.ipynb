```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

# Parameters
mu, sigma = 3, 1  # mean and standard deviation

# Generate data
x = np.linspace(-5, 5, 1000) + mu
y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# Plot
plt.figure()
plt.plot(x, y, label=f'μ={mu}, σ={sigma}')
# plt.title('Gaussian Distribution')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-0.5, 2.5, num=50)

d1 = 5 * (x - 1)**2
d2 = d1 - np.cos(10 * (x - 1))
d2 = d2 - d2.min()

plt.figure()
plt.plot(x, d1, label="Convex function")
plt.plot(x, d2, label="Non-convex function")
plt.legend()
plt.show()
```
