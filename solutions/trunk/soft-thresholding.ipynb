```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
```

```python
vx = [0, 1]
vy = [0, 1]

zx = [0, 0.7]
zy = [0, 0.7]

plt.figure()
plt.plot(vx, vy, linewidth=2, label=r"$\mathbf{v}$")
plt.plot(zx, zy, linewidth=2, label=r"$\mathbf{z}$")
plt.grid()
plt.axis("equal")
plt.xlim((-0.1, 1.1))
plt.ylim((-0.1, 1.1))
plt.xlabel(r"$\mathbf{v}_{\mathrm{x}, i}$")
plt.ylabel(r"$\mathbf{v}_{\mathrm{y}, i}$")
plt.legend()
```
